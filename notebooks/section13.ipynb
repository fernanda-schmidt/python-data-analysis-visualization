Bar Plots

```python
plants = ['Spinach', 'Turnip', 'Rhubarb', 'Broccoli', 'Kale']
died = [10,25,5,30,21]
germinated = [74,88,56,69,59]
```

```python
plt.figure(figsize=(6,6))
plt.bar(plants, germinated)
plt.title('Germination Rates')
plt.ylabel('No Of Seedlings')
```

```python
plt.figure(figsize=(6,6))
plt.bar(plants, died, color='Red')
plt.title('Failure Rates')
```

```python
plt.figure(figsize=(6,6))
plt.bar(plants, died, color='red', bottom=germinated, label='died')
plt.bar(plants, germinated, color='green', label='germinated')
plt.legend()
```

```python
plt.figure(figsize=(6, 6))
plt.bar(plants, died, color='red', label='died')
plt.bar(plants, germinated, width=0.4, color='green', label='germinated', align='edge')
plt.legend()
```

```python
plt.figure()
plt.barh(plants, germinated, color='green')
plt.barh(plants, died, color='red')
```

```python
import pandas as pd
```

```python
titanic = pd.read_csv('../datasets/data/titanic.csv')
titanic['age'] = pd.to_numeric(titanic.age, errors='coerce')
```

```python
plt.hist(titanic.age, bins=20)
```

```python
first_class = titanic[titanic.pclass == 1].age
second_class = titanic[titanic.pclass == 2].age
third_class = titanic[titanic.pclass == 3].age
plt.hist(first_class, label='1st Class', bins=20)
plt.hist(second_class, label='2nd Class', bins=20)
plt.hist(third_class, label='3rd Class', bins=20)
plt.legend()
```
