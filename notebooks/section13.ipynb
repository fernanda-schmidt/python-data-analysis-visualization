```python
nums = np.arange(5)
nums
```

```python
plt.plot(nums, nums)
plt.plot(nums, nums*nums)
plt.plot(nums, nums*nums, nums**3)
```

```python
plt.figure()
plt.plot(nums, nums)
plt.figure()
plt.plot(nums, nums*nums)
plt.figure()
plt.plot(nums, nums*nums, nums**3)
```

```python
plt.figure(figsize=(8,6))
plt.plot(nums, nums**3)
```

```python
plt.plot(nums, nums)
plt.plot(nums, nums*nums)
plt.plot(nums, nums*nums, nums**3)
```

```python
plt.plot(nums, nums, c='teal', linewidth=4)
plt.plot(nums, nums**2, c='#ff6b6b', linewidth=4)
plt.plot(nums, nums*nums, nums**3, c='#ababab', linewidth='3')
```

```python
plt.plot(nums, nums, c='teal', linewidth=4, linestyle='--')
plt.plot(nums, nums**2, c='#ff6b6b', linewidth=4, linestyle=':')
plt.plot(nums, nums*nums, nums**3, c='#ababab', linewidth='3', linestyle='-.')
```

```python
plt.plot(nums, nums, c='teal', linestyle='--', marker='>')
plt.plot(nums, nums**2, c='#ff6b6b', linestyle=':', marker='*')
plt.plot(nums, nums*nums, nums**3, c='#ababab', linestyle='-.')
```

```python
salaries = [55000, 65000, 72000, 90000, 115000, 150000]
ages = [20, 25, 30, 32, 40, 45]

plt.plot(ages, salaries)
plt.title('Company Salaries', loc='right')
plt.xlabel('Employee Age')
plt.ylabel('Avg Salary', labelpad=10)
```

```python
salaries = [55000, 65000, 72000, 90000, 115000, 150000]
ages = [20, 25, 30, 32, 40, 45]

plt.plot(ages, salaries)
plt.title('Company Salaries', loc='right')
plt.xlabel('Employee Age')
plt.ylabel('Avg Salary', labelpad=10)
plt.xticks([20,25,30,35,40,45,50])
```

```python
plt.figure(figsize=(8,8))
salaries = [55000, 65000, 72000, 90000, 115000, 150000]
ages = [20, 25, 30, 32, 40, 45]

plt.plot(ages, salaries)
plt.title('Company Salaries', loc='right')
plt.xlabel('Employee Age')
plt.ylabel('Avg Salary', labelpad=10)
plt.xticks([20, 25,30,35,40,45,50])
plt.yticks([60000,80000,100000,120000, 140000], labels=['60k', '80k', '100k', '120k', '140k'])
```

```python
plt.figure(figsize=(8, 8))
salaries = [55000, 65000, 72000, 90000, 115000, 150000]
ages = [20, 25, 30, 32, 40, 45]

plt.plot(ages, salaries)
plt.xlim(25,40)
plt.ylim(100000,120000)
```

```python
plt.figure(figsize=(5,5))
plt.plot(nums, c='teal', label='x')
plt.plot(nums**2, c='red', label='x squared')
plt.plot(nums**3, c='black', label='x cubed')
plt.legend()
```
