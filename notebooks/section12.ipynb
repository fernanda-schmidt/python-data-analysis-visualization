```python
ufos['posted'] = pd.to_datetime(ufos['posted'])
ufos.info()
```

```python
ufos.posted - ufos.date_time
```

```python
ufos['time_to_report'] = ufos.posted - ufos.date_time
ufos.head()
```

```python
ufos.sort_values('time_to_report', ascending=False)
```

```python
longest_wait = ufos.sort_values('time_to_report', ascending=False).head(10)
longest_wait['time_to_report'].dt.days / 365
```

```python
ufos['days_to_report'] = ufos['time_to_report'].dt.days
ufos.head()
```
